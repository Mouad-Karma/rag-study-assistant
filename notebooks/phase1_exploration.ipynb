```python
import os, shutil

os.makedirs("data", exist_ok=True)

for filename in uploaded.keys():
    shutil.move(filename, f"data/cloud_course")

print(os.listdir("data"))
```

```python
import fitz

# open the pdf
pdf_filename = os.listdir("data")[0]
doc = fitz.open(f"data/cloud_course")

# read all the text, page by page
text = ""
for page in doc:
    text += page.get_text()

doc.close()

print(text[1001:1500])  # show first 500 characters just to check it worked
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_text(text)

print(f"Number of chunks: {len(chunks)}")
print("\n--- Chunk 1 ---")
print(chunks[5])
print("\n--- Chunk 2 ---")
print(chunks[6])
```
